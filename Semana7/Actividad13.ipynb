```python
import numpy as np
import matplotlib.pyplot as plt

N_val = 10000
def distribucion(R=1.0, N=N_val, k=4):
    u = np.random.rand(N)
    v = np.random.rand(N)

    # Distribución del disco
    r_ok = R * np.sqrt(u)
    th_ok = 2*np.pi * v

    x_ok = r_ok * np.cos(th_ok)
    y_ok = r_ok * np.sin(th_ok)

    # Distribución de la rosa
    x_rosa, y_rosa = np.array([]), np.array([])
    count = 0

    for i in range(len(r_ok)):

      if r_ok[i] <= R*np.cos(k*th_ok[i]) and np.cos(k*th_ok[i]) > 0:
        count += 1
        x_rosa = np.append(x_rosa, x_ok[i])
        y_rosa = np.append(y_rosa, y_ok[i])

    plt.figure(figsize=(6,6))
    plt.scatter(x_ok,  y_ok,  s=6)
    plt.scatter(x_rosa,  y_rosa,  s=6)
    plt.show()

    return x_rosa, y_rosa, count

x_rosa, y_rosa, Nacc = distribucion()

R = 1
A_half = np.pi*R**2*(Nacc/N_val)
A_total = A_half*2

A_total #Área total

p_med = Nacc/N_val

error_half = np.pi*R*np.sqrt((p_med*(1-p_med))/N_val)
error_total = error_half*2

error_total
```
