## Problema 1

a) Considere la función de una flor, dada por la ecuación en coordenadas polares,

$$
r(\theta) = \frac{1}{2}(\cos{6\theta} + 1)
$$

Grafique la función de la flor, pista usar la función

```
plt.polar
```

````python
# tu codigo aqui

# Angulo
theta = np.linspace(0, 2*np.pi, 500)

# Funcion radial
r = 0.5*(np.cos(6*theta) + 1)

# Grafica polar
plt.figure()
plt.polar(theta, r)

plt.show()
````

b) Un diferencial de la flor se puede considerar como un $\Delta A$, dado por,

$$
\Delta A = \frac{\Delta \theta}{2}\Big[\frac{r_i+r_{i+1}}{2}\Big]^2
$$

Considerando ese diferencial de area, halle e imprima el area de la flor tomando 500 pasos para el ángulo $\theta$

Pista si genera un `linspace`, ponga `retstep = True`



````python
flower_area = 0
## Tu codigo aqui


# 500 pasos
theta, dtheta = np.linspace(0, 2*np.pi, 500, retstep=True)

# radios
r = 0.5*(np.cos(6*theta) + 1)

# suma de diferenciales
for i in range(len(theta)-1):

    r_prom = (r[i] + r[i+1]) / 2

    dA = (dtheta/2) * (r_prom**2)

    flower_area += dA

flower_area

## Codigo para verificar el resultado no modificar
flower_area
````

````python
flower_area = 0
## Tu codigo aqui

# limites
a = 0
b = 2*np.pi

# funcion r(theta)
r = lambda th: 0.5*(np.cos(6*th) + 1)

# funcion a integrar
f = lambda th: 0.5*(r(th)**2)

# numero de trapecios
N = 500
theta = np.linspace(a, b, N+1)

# paso
h = theta[1] - theta[0]

# regla del trapecio
flower_area += 0.5*(f(theta[0]) + f(theta[-1]))

for i in range(1, len(theta)-1):
    flower_area += f(theta[i])

flower_area *= h

## Codigo para verificar el resultado no modificar
flower_area
````

P1b) Argumente por qué el diferencial de area está dado por la expresión de arriba.

Tu texto aquí:

El área en coordenadas polares está dada por A = 1/2 ∫r**2 (theta) dtheta. Al poner el intervalo angular en pequeños intervalos iguales se aproxima la integral mediante el método de los trapecios, donde cada lugar circular se aproxima por un trapecio en la variable angular.

## Problema 2

Considere una esfera de radio $2$ , el diferencial de volumen en coordenadas esfericas está dado por,

$$
dV = \rho^2\sin{\phi} \,d\rho \,d\theta \,d\phi
$$

El diferencial de volumen se puede simplificar para el caso de la esfera a,

$$
\Delta V = \frac{\rho^3}{3}\sin{\phi} \,d\theta \,d\phi
$$

a) Calcule e imprima el volumen de una esfera de radio 2, con 500 pasos para cada coordenada.

````python
sphere_volume = 0
## Tu codigo aqui

r = 2

# limites
a_phi = 0
b_phi = np.pi

a_theta = 0
b_theta = 2*np.pi

N = 500

# funcion
f = lambda phi: (r**3/3)*np.sin(phi)

phi = np.linspace(a_phi, b_phi, N+1)
h_phi = phi[1] - phi[0]

# trapecio en phi
integral_phi = 0
integral_phi += 0.5*(f(phi[0]) + f(phi[-1]))

for i in range(1,len(phi)-1):
    integral_phi += f(phi[i])

integral_phi *= h_phi

# ahora integrar en theta
theta = np.linspace(a_theta, b_theta, N+1)
h_theta = theta[1] - theta[0]

sphere_volume = integral_phi * (b_theta - a_theta)


## Codigo para verificar el resultado no modificar
sphere_volume
````

b) Justifique por qué a partir del diferencial en coordenadas esfericas se puede obtener el diferencial $\Delta V$ anterior.

El diferencial de volumen en coordenadas esféricas es dV=ρ2sinϕdρdθdϕ.
Integrando primero en la coordenada radial desde 0 hasta 𝑅 se obtiene R**3/3. El volumen restante depende solo de los ángulos y puede evaluarse mediante el método del trapecio, aproximando la integral angular como suma de trapecios.
