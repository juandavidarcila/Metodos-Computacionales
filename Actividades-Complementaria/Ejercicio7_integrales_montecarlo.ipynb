# Integral de radiación

La potencia (por unidad de área) emitida por un cuerpo a una temperatura $T$, para cada frecuencia de luz $\nu$, se rige por la ley de Planck:
$$
I(\nu) = \frac{2\pi h \nu^3}{c^2} \Big/ \left(\exp\left(\frac{h \nu}{k T}\right)-1 \right)
$$
Donde $k$ es la constante de Boltzmann, $h$ es la constante de Planck, y $c$ es la velocidad de la luz.

La potencia total será la integral sobre todas las frecuencias $\nu$:
$$
P = \int_0^\infty I(\nu) \;d \nu
$$

Debe usar una sustitución para sacar la temperatura y el resto de variables del integrando, tal que quede una integral sin ninguna constante física adentro.

$$
P = \text{constantes} \times T^4 \times \int_0^\infty \left( \cdots\right) \text{d} u
$$

Esta constante tiene el nombre de Stefan-Boltzmann.

### Punto 1

Use `scipy.quad` para obtener el valor (y la incertidumbre) de esta constante. Reemplace los valores de las constantes físicas ya mencionadas y compare con el verdadero valor de la constante de Stefan-Boltzmann: $5.6703744191 \times 10^{−8} \frac{W}{m^2 K^4}$

```python
# constantes físicas
k = 1.380649e-23
h = 6.62607015e-34
c = 2.99792458e8

def Int(u):
    return u**3/(np.exp(u)-1)

I, error = quad(Int,0,np.inf)
const = (2*np.pi*k**4)/(h**3*c**2)
sigma = const*I

print("Integral =",I)
print("Error =",error)
print("Sigma calculada =",sigma)
print("Sigma real =",5.6703744191e-8)
```

### Punto 2

Si la distribución que usamos para los números aleatorios no es sólo la uniforme, sino alguna más general, digamos $p(x)$, la integral puede evaluarse como

$$
\int_D f(x) \, dx \approx \frac{1}{N} \sum_{i=1}^N \frac{f(x_i)}{p(x_i)}
$$

Donde los puntos $x_i$ son muestreados según la distribución $p$.

Use la distribución exponencial, que tiene dominio $D=[0,\infty)$, para realizar esta integral. Debería obtener un valor similar si usa $N\approx 10\,000$ puntos.

```python
N = 10000

# muestreo exponencial
u = random.exponential(scale=1,size=N)

f = u**3/(np.exp(u)-1)
p = np.exp(-u)

I_mc = np.mean(f/p)
sigma_mc = const*I_mc

print("Integral Monte Carlo =",I_mc)
print("Sigma Monte Carlo =",sigma_mc)
```
