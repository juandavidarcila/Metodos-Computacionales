```python
import numpy as np

# Parámetros
a = 1.0
k1 = 10.0
k2 = 30.0
L1 = 1.0
L2 = 2.0
m = 1.0
g = 9.8

# Sistema de ecuaciones
def F(X):
    x, y = X

    l1 = np.sqrt((x+a)**2 + y**2)
    l2 = np.sqrt((x-a)**2 + y**2)

    f1 = -k1*(l1-L1)*(x+a)/l1 - k2*(l2-L2)*(x-a)/l2
    f2 = -k1*(l1-L1)*(y)/l1 - k2*(l2-L2)*(y)/l2 - m*g

    return np.array([f1, f2])

# Jacobiano numérico
def Jacobiano(F, X, h=1e-5):
    n = len(X)
    J = np.zeros((n,n))

    for i in range(n):
        X1 = X.copy()
        X2 = X.copy()

        X1[i] += h
        X2[i] -= h

        J[:,i] = (F(X1) - F(X2)) / (2*h)

    return J

# Método de Newton
def Newton(F, X0, tol=1e-6, max_iter=100):
    X = X0

    for i in range(max_iter):
        J = Jacobiano(F, X)
        dX = np.linalg.solve(J, -F(X))
        X = X + dX

        if np.linalg.norm(dX) < tol:
            print("Convergió en", i, "iteraciones")
            return X

    print("No convergió")
    return X

# Semilla inicial
X0 = np.array([0.0, -1.0])

sol = Newton(F, X0)
print("Solución:", sol)
```
