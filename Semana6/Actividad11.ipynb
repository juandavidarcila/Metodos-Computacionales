```python
def GetNewtonMethod(f,df,xn,itmax = 1000, precision=1e-14):

    error = 1.
    it = 0

    while error >= precision and it < itmax:

      dfx = df(xn)

      # Evitar división por cero
      if np.abs(dfx) < 1e-16:
            print("Derivada cercana a cero. Método falla.")
            return False

      xn1 = xn - f(xn)/df(xn)
      error = np.abs( xn1-xn )

      xn  = xn1
      it += 1


    if it == itmax:
        return False
    else:
        return xn
```

```python
def GetAllRoots(f, df, x, tolerancia_decimales=8, tol_unicidad=1e-6):

    Roots = []

    for x0 in x:

        root = GetNewtonMethod(f, df, x0)

        if root is not False:

            # Redondeamos
            croot = np.round(root, tolerancia_decimales)

            # Verificamos que no esté ya (con tolerancia)
            if not any(np.isclose(croot, r, atol=tol_unicidad) for r in Roots):
                Roots.append(croot)

    Roots = np.array(Roots)
    Roots.sort()

    return Roots
```

```python
def GetLegendre(n):

  x = sym.Symbol('x',Real=True)
  y = sym.Symbol('y',Real=True)

  y = (x**2 - 1)**n

  p = sym.diff(y,x,n)/(2**n * factorial(n))

  return p
```

```python
def GetLaguerre(n):
  x = sym.Symbol('x',Real=True)
  y = sym.Symbol('y',Real=True)

  y = sym.exp(-x)*x**n

  p = sym.diff(y,x,n)*(sym.exp(x)/factorial(n))

  return p.simplify()
```

```python
GetLaguerre(1)
```

```python
Laguerre = []
DLaguerre = []

x = sym.Symbol('x',Real=True)
n=7

for i in range(n+1):

    poly = GetLaguerre(i)

    Laguerre.append(poly)
    DLaguerre.append(sym.diff(poly,x))
```

```python
def GetRootsPolynomial(n,xi,poly,dpoly):

    x = sym.Symbol('x',Real=True)
    pn = sym.lambdify([x],poly[n],'numpy')
    dpn = sym.lambdify([x],dpoly[n],'numpy')

    Roots = GetAllRoots(pn,dpn,xi)

    return Roots
```

```python
xi = np.linspace(0,40,100,dtype=np.longdouble)
Roots = GetRootsPolynomial(n,xi,Laguerre,DLaguerre)
Roots
```
